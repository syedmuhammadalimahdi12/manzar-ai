```python
# Step 1: Install ultralytics
!pip install ultralytics -q

# Step 2: Write a corrected data.yaml for Kaggle's paths
dataset_root = "/kaggle/input/datasets/abdullahmirza2004/manzar-final-dataset/manzar_final_dataset"

yaml_content = f"""
path: {dataset_root}
train: images/train
val: images/val
nc: 4
names: ['skid_mark', 'vehicle_damage', 'debris_field', 'point_of_impact']
"""

with open("/kaggle/working/data.yaml", "w") as f:
    f.write(yaml_content)

print("data.yaml written:")
print(yaml_content)

# Step 3: Train YOLOv8s
from ultralytics import YOLO

model = YOLO('yolov8s.pt')  # pretrained base model, we fine-tune from here

results = model.train(
    data="/kaggle/working/data.yaml",
    epochs=80,
    imgsz=960,
    batch=8,
    project="/kaggle/working/manzar_runs",
    name="yolov8s_manzar_v3",
    patience=20,
    cos_lr=True,        # naya — smoother convergence
    save_period=10,      # har 10 epochs pe checkpoint save karega, safety ke liye
)
```
